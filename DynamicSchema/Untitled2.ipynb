```python
import logging
from typing import List, Dict, Any
import openrouter
import re

# Configure logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger("GeminiAgent")

class GeminiAgent:
    def __init__(
        self,
        system_prompt: str,
        api_key: str,
        model_name: str = "google/gemini-1.5-pro",
    ):
        """
        Initializes the GeminiAgent with the specified system prompt and OpenRouter configurations.

        Args:
            system_prompt (str): The system-level instruction for the AI model.
            api_key (str): The API key for OpenRouter.
            model_name (str, optional): Name of the OpenRouter model to use. Defaults to "openrouter-gemini-1.5".
        """
        self.api_key = api_key
        self.model_name = model_name
        self.system_prompt = system_prompt

        # Initialize OpenRouter
        self.client = openrouter.OpenRouterClient(api_key=api_key)
        logger.info("OpenRouter initialized.")

        # Load the model
        self.model = self.client.get_model(model_name)
        logger.info(f"Model '{model_name}' loaded.")

        self.chat_session: openrouter.ChatSession = None
        logger.info("GeminiAgent initialized.")

    def start_session(self):
        """
        Starts a new chat session with the AI model.
        """
        self.chat_session = self.model.start_chat_session(self.system_prompt)
        logger.info("Chat session started with system prompt.")

    def parse_content_blocks(self, blocks: List[Dict[str, Any]]) -> (List[str], str):
        """
        Parses the response content blocks to extract text.

        Args:
            blocks (List[Dict[str, Any]]): The list of content blocks from the model.

        Returns:
            Tuple containing:
                - List of reasoning steps (strings).
                - Final answer string.
        """
        reasoning_steps = []
        final_answer = ""

        for block in blocks:
            text = block.get("text", "")
            thinking_text = re.findall(r"<thinking>(.*?)</thinking>", text, re.DOTALL)
            reasoning_steps.extend(thinking_text)
            if "<final_answer>" in text:
                final_answer = re.findall(r"<final_answer>(.*?)</final_answer>", text, re.DOTALL)[0]

        return reasoning_steps, final_answer

    def prompt(self, user_message: str) -> str:
        """
        Sends a user message to the AI model and returns the response.

        Args:
            user_message (str): The message from the user.

        Returns:
            str: The AI model's response.
        """
        if self.chat_session is None:
            raise ValueError("Please start a session before prompting the agent.")

        try:
            response = self.chat_session.send_message(user_message)
            logger.info(f"User: {user_message}")
            logger.debug(f"Raw Response: {response}")
            
            # Extract the text from the response
            response_text = response.get("text", "")
            logger.info(f"Assistant: {response_text}")
            return response_text
        except Exception as e:
            logger.error(f"Error sending message: {e}")
            return f"Error sending message: {e}"

import asyncio

async def generate_cot_structure(self, user_message: str):
    """
    Generates a chain-of-thought structure for the given user message.

    Args:
        user_message (str): The message from the user.

    Yields:
        dict: The generated chain-of-thought items.
    """
    if self.chat_session is None:
        raise ValueError("Please start a session before generating a chain-of-thought structure.")

    try:
        # Send the user message to the model
        response = self.chat_session.send_message(user_message)
        logger.info(f"User: {user_message}")
        logger.debug(f"Raw Response: {response}")

        # Extract the reasoning steps and final answer
        reasoning_steps, final_answer = self.parse_content_blocks(response.get("content", []))

        # Yield the chain-of-thought structure
        yield {"type": "cot_structure", "content": f"<thinking>\n{reasoning_steps}\n</thinking>"}

        # Yield each reasoning step
        for step in reasoning_steps:
            yield {"type": "step", "content": f"Thinking about: {step}"}
            # Simulate processing time for each step
            await asyncio.sleep(1)

        # Yield the final answer
        yield {"type": "response", "content": final_answer}

    except Exception as e:
        logger.error(f"Error generating content: {e}")
        yield {"type": "response", "content": f"Error generating content: {e}"}

async def generate_content(self, user_message: str):
    """
    Generates content based on the user message and yields the response.

    Args:
        user_message (str): The message from the user.

    Yields:
        str: The generated content.
    """
    if self.chat_session is None:
        raise ValueError("Please start a session before generating content.")

    try:
        # Send the user message to the model
        response = self.chat_session.send_message(user_message)
        logger.info(f"User: {user_message}")
        logger.debug(f"Raw Response: {response}")

        # Extract the text from the response
        response_text = response.get("text", "")
        logger.info(f"Assistant: {response_text}")
        yield response_text

    except Exception as e:
        logger.error(f"Error generating content: {e}")
        yield f"Error generating content: {e}"
        
import os
a = GeminiAgent(system_prompt='', api_key=os.getenv("_API_KEY"))
```

```python
import logging
from typing import List, Dict, Any
import re

import requests
import logging
from typing import List, Dict, Any, Optional, Union
import json

class OpenRouterClient:
    """
    A client to interact with the OpenRouter API.
    """
    def __init__(self, api_key: str, site_url: Optional[str] = None, site_name: Optional[str] = None):
        self.api_key = api_key
        self.site_url = site_url
        self.site_name = site_name
        self.endpoint = "https://openrouter.ai/api/v1/chat/completions"
        self.headers = {
            "Authorization": f"Bearer {self.api_key}",
            "Content-Type": "application/json",
            "Accept": "text/event-stream"  # Ensure Accept header is set for streaming
        }
        if self.site_url:
            self.headers["HTTP-Referer"] = self.site_url
        if self.site_name:
            self.headers["X-Title"] = self.site_name

    def create_completion(
        self,
        model: Optional[str],
        messages: List[Dict[str, Any]],
        max_completion_tokens: Optional[int] = None,
        temperature: Optional[float] = None,
        top_p: Optional[float] = None,
        stream: bool = False,
        tools: Optional[List[Dict[str, Any]]] = None,
        tool_choice: Optional[Union[str, Dict[str, Any]]] = None,
        stop: Optional[Union[str, List[str]]] = None,
        n: int = 1
    ) -> Dict[str, Any]:
        """
        Create a completion using OpenRouter API.

        Parameters:
            model (str): Model name to use.
            messages (list): List of message dictionaries.
            max_completion_tokens (int): Maximum number of tokens to generate.
            temperature (float): Sampling temperature.
            top_p (float): Nucleus sampling parameter.
            stream (bool): Whether to enable streaming.
            tools (list): List of tools for tool calling.
            tool_choice (str or dict): Tool choice parameter.
            stop (str or list): Stop sequences.

        Returns:
            dict: API response.
        """
        payload = {
            "messages": messages,
        }

        if model:
            payload["model"] = model
        if max_completion_tokens:
            payload["max_completion_tokens"] = max_completion_tokens
        if temperature is not None:
            payload["temperature"] = temperature
        if top_p is not None:
            payload["top_p"] = top_p
        if stream:
            payload["stream"] = True
        if tools:
            payload["tools"] = tools
        if tool_choice:
            payload["tool_choice"] = tool_choice
        if stop:
            payload["stop"] = stop

        logger.debug(f"OpenRouter Payload: {json.dumps(payload, indent=2)}")

        try:
            response = requests.post(
                self.endpoint,
                headers=self.headers,
                json=payload,
                stream=stream
            )
        except requests.exceptions.RequestException as e:
            logger.error(f"Request Exception: {e}")
            raise Exception(f"Request Exception: {e}")

        if response.status_code != 200:
            logger.error(f"OpenRouter request failed with status {response.status_code}: {response.text}")
            try:
                error_response = response.json()
                logger.debug(f"Full response: {json.dumps(error_response, indent=2)}")
            except json.JSONDecodeError:
                logger.debug("Response is not in JSON format.")
            raise Exception(f"OpenRouter API request failed with status {response.status_code}: {response.text}")

        if stream:
            return response
        else:
            try:
                return response.json()
            except json.JSONDecodeError:
                logger.error("Failed to decode JSON response.")
                raise Exception("Failed to decode JSON response.")


# Configure logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger("GeminiAgent")

class GeminiAgent:
    def __init__(
        self,
        system_prompt: str,
        api_key: str,
        model_name: str = "google/gemini-1.5-pro",
    ):
        """
        Initializes the GeminiAgent with the specified system prompt and OpenRouter configurations.

        Args:
            system_prompt (str): The system-level instruction for the AI model.
            api_key (str): The API key for OpenRouter.
            model_name (str, optional): Name of the OpenRouter model to use. Defaults to "openrouter-gemini-1.5".
        """
        self.api_key = api_key
        self.model_name = model_name
        self.system_prompt = system_prompt

        # Initialize OpenRouter
        self.client = openrouter.OpenRouterClient(api_key=api_key)
        logger.info("OpenRouter initialized.")

        # Load the model
        self.model = self.client.get_model(model_name)
        logger.info(f"Model '{model_name}' loaded.")

        self.chat_session: openrouter.ChatSession = None
        logger.info("GeminiAgent initialized.")

    def start_session(self):
        """
        Starts a new chat session with the AI model.
        """
        self.chat_session = self.model.start_chat_session(self.system_prompt)
        logger.info("Chat session started with system prompt.")

    def parse_content_blocks(self, blocks: List[Dict[str, Any]]) -> (List[str], str):
        """
        Parses the response content blocks to extract text.

        Args:
            blocks (List[Dict[str, Any]]): The list of content blocks from the model.

        Returns:
            Tuple containing:
                - List of reasoning steps (strings).
                - Final answer string.
        """
        reasoning_steps = []
        final_answer = ""

        for block in blocks:
            text = block.get("text", "")
            thinking_text = re.findall(r"<thinking>(.*?)</thinking>", text, re.DOTALL)
            reasoning_steps.extend(thinking_text)
            if "<final_answer>" in text:
                final_answer = re.findall(r"<final_answer>(.*?)</final_answer>", text, re.DOTALL)[0]

        return reasoning_steps, final_answer

    def prompt(self, user_message: str) -> str:
        """
        Sends a user message to the AI model and returns the response.

        Args:
            user_message (str): The message from the user.

        Returns:
            str: The AI model's response.
        """
        if self.chat_session is None:
            raise ValueError("Please start a session before prompting the agent.")

        try:
            response = self.chat_session.send_message(user_message)
            logger.info(f"User: {user_message}")
            logger.debug(f"Raw Response: {response}")
            
            # Extract the text from the response
            response_text = response.get("text", "")
            logger.info(f"Assistant: {response_text}")
            return response_text
        except Exception as e:
            logger.error(f"Error sending message: {e}")
            return f"Error sending message: {e}"

import asyncio

async def generate_cot_structure(self, user_message: str):
    """
    Generates a chain-of-thought structure for the given user message.

    Args:
        user_message (str): The message from the user.

    Yields:
        dict: The generated chain-of-thought items.
    """
    if self.chat_session is None:
        raise ValueError("Please start a session before generating a chain-of-thought structure.")

    try:
        # Send the user message to the model
        response = self.chat_session.send_message(user_message)
        logger.info(f"User: {user_message}")
        logger.debug(f"Raw Response: {response}")

        # Extract the reasoning steps and final answer
        reasoning_steps, final_answer = self.parse_content_blocks(response.get("content", []))

        # Yield the chain-of-thought structure
        yield {"type": "cot_structure", "content": f"<thinking>\n{reasoning_steps}\n</thinking>"}

        # Yield each reasoning step
        for step in reasoning_steps:
            yield {"type": "step", "content": f"Thinking about: {step}"}
            # Simulate processing time for each step
            await asyncio.sleep(1)

        # Yield the final answer
        yield {"type": "response", "content": final_answer}

    except Exception as e:
        logger.error(f"Error generating content: {e}")
        yield {"type": "response", "content": f"Error generating content: {e}"}

async def generate_content(self, user_message: str):
    """
    Generates content based on the user message and yields the response.

    Args:
        user_message (str): The message from the user.

    Yields:
        str: The generated content.
    """
    if self.chat_session is None:
        raise ValueError("Please start a session before generating content.")

    try:
        # Send the user message to the model
        response = self.chat_session.send_message(user_message)
        logger.info(f"User: {user_message}")
        logger.debug(f"Raw Response: {response}")

        # Extract the text from the response
        response_text = response.get("text", "")
        logger.info(f"Assistant: {response_text}")
        yield response_text

    except Exception as e:
        logger.error(f"Error generating content: {e}")
        yield f"Error generating content: {e}"
        
import os
a = GeminiAgent(system_prompt='', api_key=os.getenv("GOOGLE_AI_API_KEY"))
```

```python
import os
import json
import requests
import sseclient

# Initialize Anthropic Client with API Key
API_KEY = os.getenv("ANTHROPIC_API_KEY")  # Ensure your API key is set in environment variables
API_URL = "https://api.anthropic.com/v1/messages"

# Define available tools (functions) that the model can call
def get_weather(location):
    """Simulate fetching weather information for a given location."""
    # In a real implementation, integrate with a weather API
    return f"The weather in {location} is 72°F and sunny."

# Map function names to actual functions
TOOLS = {
    "get_weather": get_weather
}

def call_tool(function_name, function_input):
    """Call the appropriate tool based on function_name."""
    tool = TOOLS.get(function_name)
    if tool:
        return tool(**function_input)
    else:
        return f"Error: Unknown function '{function_name}'."

class ClaudeStreamer:
    def __init__(self, api_key, api_url):
        self.api_key = api_key
        self.api_url = api_url
        self.headers = {
            'x-api-key': self.api_key,
            'Content-Type': 'application/json',
            'anthropic-version': '2023-06-01'
        }
        # Initialize parser state
        self.current_block = None
        self.buffer = ""
    
    def response_to_model(self, result, tool_use_id):
        """Send the result of a tool call back to the model."""
        # Prepare the response payload
        response_data = {
            "type": "tool_result",
            "tool_use_id": tool_use_id,
            "content": result
        }
        # In a real scenario, you would send this response back to the Anthropic API
        # For demonstration, we'll just print it
        print(f"\nSending result back to model: {result}")
        # Example: Send response_data to the API if supported
        # response = requests.post(self.api_url, headers=self.headers, json=response_data)
        # Handle response as needed

    def handle_event(self, event):
        """Handle incoming events from Claude's streaming response."""
        try:
            event_data = json.loads(event.data)
        except json.JSONDecodeError:
            print(f"Failed to parse event data: {event.data}")
            return

        event_type = event_data.get('type')
        
        if event_type == 'message_start':
            model = event_data['message'].get('model', 'unknown')
            print(f"Message start: Model - {model}")
        
        elif event_type == 'content_block_start':
            self.current_block = event_data['content_block'].get('type', 'unknown')
            print("\n--- Starting content block ---")
        
        elif event_type == 'ping':
            print("\nPing received (keep-alive)...")
        
        elif event_type == 'content_block_delta':
            delta_text = event_data['delta'].get('text', '')
            self.buffer += delta_text
            print(delta_text, end='', flush=True)
        
        elif event_type == 'content_block_stop':
            print("\n--- Content block finished ---\n")
            self.current_block = None
            self.buffer = ""
        
        elif event_type == 'message_stop':
            print("\n--- Message finished ---\n")
        
        elif event_type == 'message_delta':
            delta = event_data.get('delta', {})
            stop_reason = delta.get('stop_reason')
            if stop_reason == 'tool_use':
                tool_name = delta.get('tool', {}).get('name')
                tool_input = delta.get('tool', {}).get('input')
                tool_id = delta.get('tool', {}).get('id')
                if tool_name and tool_input:
                    print(f"\nFunction call detected: {tool_name} with input: {tool_input}")
                    # Execute the tool function
                    result = call_tool(tool_name, tool_input)
                    # Respond back to the model with the tool result
                    self.response_to_model(result, tool_id)
            else:
                print(f"\nMessage delta received: {delta}")
        
        elif event_type == 'tool_use':
            # Deprecated or alternative handling
            tool_name = event_data.get('name')
            tool_input = event_data.get('input')
            tool_id = event_data.get('id')
            if tool_name and tool_input:
                print(f"\nFunction call detected: {tool_name} with input: {tool_input}")
                # Execute the tool function
                result = call_tool(tool_name, tool_input)
                # Respond back to the model with the tool result
                self.response_to_model(result, tool_id)
        
        else:
            print(f"Unknown event type: {event_type}")
    
    def stream_message(self, message_content, model="claude-3-5-sonnet-20240620", max_tokens=1024):
        """Stream a message to Claude's API and handle streaming events."""
        # Define available tools for the model
        tools = [
            {
                "name": "get_weather",
                "description": "Get the current weather in a given location",
                "input_schema": {
                    "type": "object",
                    "properties": {
                        "location": {
                            "type": "string",
                            "description": "The city and state, e.g., San Francisco, CA"
                        }
                    },
                    "required": ["location"]
                }
            }
        ]
        
        # Create the request payload for streaming with tools enabled
        data = {
            "model": model,
            "messages": [{"role": "user", "content": message_content}],
            "max_tokens": max_tokens,
            "stream": True,  # Enable streaming for real-time updates
            "tools": tools,  # Provide tool options to the model
            "tool_choice": {"type": "any"}  # Allow the model to choose any available tool
        }
        
        # Make the POST request with streaming enabled
        try:
            response = requests.post(self.api_url, headers=self.headers, json=data, stream=True)
            response.raise_for_status()
        except requests.exceptions.RequestException as e:
            print(f"Request failed: {e}")
            return
        
        # Use the SSE client to handle the event stream
        client = sseclient.SSEClient(response)
        
        print("\n--- Starting to stream response from Claude ---\n")
        try:
            for event in client.events():
                if event.data:
                    self.handle_event(event)
        except Exception as e:
            print(f"Error while streaming: {e}")
        finally:
            print("\n--- Streaming ended ---\n")

# Example usage: Ask the model for weather information
if __name__ == "__main__":
    streamer = ClaudeStreamer(API_KEY, API_URL)
    streamer.stream_message("What's the weather like in San Francisco?")
```
